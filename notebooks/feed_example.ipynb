# Waze Feed Data - Basic Report

This notebook demonstrates how to use `load_feed_json` to read Waze feed data from S3 and run a basic report using the `wazeasy` package.

```python
from wazeasy import utils, reports
import yaml
```

## Load configuration

```python
with open("config_montevideo.yaml", "r") as f:
    config = yaml.safe_load(f)

timezone = config["Montevideo"]["timezone"]["timezone_name"]
s3_path = config["Montevideo"]["s3_path"]
aws_profile = config["Montevideo"]["aws_profile"]
print(f"Timezone: {timezone}")
print(f"S3 path: {s3_path}")
```

## Load feed data from S3

Use `load_feed_json` to read the JSON files collected from the Waze feeds and aggregate them into a single DataFrame.

```python
ddf = utils.load_feed_json(
    s3_path,
    start='2026-06-25',
    end='2026-06-28',
    data_type='jams',
    profile_name=aws_profile,
    use_dask=True,
    filter_level_5=True
)
print(f"Loaded {len(ddf)} records")
ddf.head()
```

## Process timestamps

Convert the `ts` column to the local timezone and add time attributes (year, month, date, hour).

```python
utils.handle_time(ddf, timezone)

# Add a dummy 'region' column required by the TCI functions
# (normally set by assign_geography_to_jams)
ddf['region'] = 'region'

ddf[['ts', 'local_time', 'date', 'hour', 'street', 'level']].head()
```

## Run Basic Report

```python
reports.run_basic_report(ddf, start_date='2026-06-25', end_date='2026-06-28')
```
